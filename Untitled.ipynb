```python
import requests
import re
import time
from bs4 import BeautifulSoup
import pandas as pd

# 浏览器头，防止被拦截
HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:120.0) "
        "Gecko/20100101 Firefox/120.0"
    ),
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8",
    "Accept-Language": "en-US,en;q=0.9",
    "Referer": "https://www.tvmaze.com/",
    "Connection": "keep-alive",
    "Upgrade-Insecure-Requests": "1",
}
PROXIES = {"http": None, "https": None}

print("✅ 环境加载完成")
```

```python
def clean_text(text):
    """去除多余空白"""
    if not text:
        return ""
    text = re.sub(r"\s+", " ", text)
    return text.strip()

def parse_show_page(url):
    """抓取并解析单个剧集详情页（适配2025新版结构）"""
    try:
        res = requests.get(url, headers=HEADERS, timeout=15, proxies=PROXIES)
        if res.status_code != 200:
            print(f"⚠️ 状态码 {res.status_code} — 跳过 {url}")
            return None
    except Exception as e:
        print(f"❌ 请求失败 {url}: {e}")
        return None

    soup = BeautifulSoup(res.text, "html.parser")

    # --- 标题 ---
    title_tag = soup.find("h1")
    title = clean_text(title_tag.get_text()) if title_tag else ""

    # --- 概要 summary ---
    summary_tag = soup.find("div", class_=lambda x: x and "summary" in x)
    summary = clean_text(summary_tag.get_text()) if summary_tag else ""

    # --- 提取基础信息 ---
    info = {"Premiered": "", "Ended": "", "Rating": "", "Genres": "", "Status": "", "Network": ""}

    for p in soup.select("section a, section p, div p"):
        text = clean_text(p.get_text(" ", strip=True))
        if ":" not in text:
            continue
        key, val = text.split(":", 1)
        key, val = key.strip(), val.strip()
        if key in info:
            info[key] = val

    # --- 兜底提取 genres ---
    if not info["Genres"]:
        genres_block = soup.find_all("a", href=re.compile("/shows\\?genre="))
        if genres_block:
            info["Genres"] = " | ".join(g.get_text(strip=True) for g in genres_block)

    return {
        "Title": title,
        "First air date": info["Premiered"],
        "End date": info["Ended"],
        "Rating": info["Rating"],
        "Genres": info["Genres"],
        "Status": info["Status"],
        "Network": info["Network"],
        "Summary": summary,
        "URL": url
    }
```

```python
print("🔍 Starting scraping ...")

all_links = []
seen_urls = set()

# 抓取前3页（每页约100剧集，可调整范围）
for page in range(3):
    list_url = f"https://www.tvmaze.com/shows?page={page}"
    print(f"Fetching list page {page}: {list_url}")
    try:
        res = requests.get(list_url, headers=HEADERS, timeout=15, proxies=PROXIES)
    except Exception as e:
        print("❌ 页面加载失败：", e)
        continue
    soup = BeautifulSoup(res.text, "html.parser")

    # 新版结构：每个剧集在 a.link-block 下
    show_links = soup.select("a.link-block[href^='/shows/']")

    if not show_links:
        print("⚠️ 没找到 show 链接，检查 HTML 结构。")
        continue

    for a in show_links:
        show_url = "https://www.tvmaze.com" + a["href"]
        if show_url not in seen_urls:
            seen_urls.add(show_url)
            all_links.append(show_url)

print(f"✅ 共发现 {len(all_links)} 个唯一剧集链接。")
```

```python
all_data = []
total = len(all_links)
print(f"开始解析 {total} 个剧集详情页...")

for i, url in enumerate(all_links, start=1):
    data = parse_show_page(url)
    if data:
        all_data.append(data)
        print(f"✅ ({i}/{total}) {data['Title']}")
    else:
        print(f"⚠️ ({i}/{total}) 解析失败: {url}")
    time.sleep(0.5)  # polite delay

print(f"📦 完成，共收集到 {len(all_data)} 条有效记录。")
```

```python
df = pd.DataFrame(all_data)
df = df.drop_duplicates(subset=["Title"], keep="first")

OUTPUT_CSV = "LaiYiming_2254030.csv"  # ⚠️ 按要求改成你的文件名
df.to_csv(OUTPUT_CSV, index=False, encoding="utf-8-sig")

print(f"✅ 已保存 {len(df)} 条唯一剧集到文件：{OUTPUT_CSV}")
```
